```python
df = orders.merge(Customers,on = "CustomerID",how = "left")
```

```python
df = df.merge(orderdetails,on = "OrderID",how = "inner")
```

```python

# Rename the remaining _x columns to remove _x
df = df.rename(columns={
    'OrderDetailID_x': 'OrderDetailID',
    'ProductID_x': 'ProductID',
    'Quantity_x': 'Quantity',
    'UnitPrice_x': 'UnitPrice',
    'DiscountPct_x': 'DiscountPct',
    'SalesAmount_x': 'SalesAmount',
    'Profit_x': 'Profit'
})
```

```python
print("df columns:", df.columns.tolist())
print("products columns:", products.columns.tolist())
```

```python
df = df.merge(
    products,
   on = "ProductID",
   how = "left")
```

```python
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["Monthname"] = df["OrderDate"].dt.month_name()
df["Year Name"] = df["OrderDate"].dt.to_period("M").astype(str)
```
